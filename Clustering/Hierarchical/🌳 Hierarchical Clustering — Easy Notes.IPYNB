# 🌳 Hierarchical Clustering — Easy Notes

## 1. What is Hierarchical Clustering?

**Hierarchical Clustering** is an **Unsupervised Machine Learning algorithm** used to group similar data points into clusters.

The main idea is:

> **Find similar points → merge them → create a hierarchy/tree → choose final clusters.**

### Simple example

Suppose we have 6 customers:

```text
A   B   C   D   E   F
```

If A and B are very similar:

```text
A + B → AB
```

Then C and D are similar:

```text
C + D → CD
```

Then:

```text
AB + CD → ABCD
```

So the algorithm keeps creating bigger groups.

---

# 2. Why is it called "Hierarchical"?

**Hierarchy means levels or layers.**

The algorithm creates clusters step-by-step.

```text
A   B   C   D   E
│   │   │   │   │
└─AB┘   │   └─DE┘
   │    │     │
   └─ABC┘    │
       │     │
       └─ABCDE
```

It creates a **tree-like structure** of clusters.

---

# 3. Type of Hierarchical Clustering

There are two main types:

## 1. Agglomerative Clustering ⭐

This is the most commonly used type.

It starts with:

```text
A   B   C   D   E
```

Every point is its own cluster.

Then it keeps **merging** similar clusters.

```text
[A] [B] [C] [D] [E]

      ↓

[AB] [C] [DE]

      ↓

[ABC] [DE]

      ↓

[ABCDE]
```

### Remember:

> **Agglomerative = Bottom → Up**

Small clusters become bigger clusters.

---

## 2. Divisive Clustering

This works in the opposite direction.

It starts with one large cluster:

```text
[ABCDE]
```

Then keeps splitting it:

```text
[ABCDE]
   ↓
[ABC] [DE]
   ↓
[AB] [C] [D] [E]
```

### Remember:

> **Divisive = Top → Down**

For beginners, focus mainly on **Agglomerative Clustering**.

---

# 4. How Agglomerative Clustering Works

The process is:

```text
1. Start with individual points
          ↓
2. Calculate distance
          ↓
3. Find closest clusters
          ↓
4. Merge them
          ↓
5. Calculate distance again
          ↓
6. Merge again
          ↓
7. Continue until one hierarchy is created
          ↓
8. Use Dendrogram to choose clusters
```

---

# 5. Step-by-Step Example

Suppose:

```text
A   B   C   D
```

### Step 1 — Every point is separate

```text
[A] [B] [C] [D]
```

---

### Step 2 — Find closest points

Suppose A and B are closest.

```text
[AB] [C] [D]
```

---

### Step 3 — Find closest clusters again

Suppose C and D are closest.

```text
[AB] [CD]
```

---

### Step 4 — Merge again

Finally:

```text
[ABCD]
```

This complete merging process creates a hierarchy.

# 6. Distance

Hierarchical Clustering needs to know:

> **How similar or close are two data points/clusters?**

It can use distance measures such as:

### Euclidean Distance

Very commonly used.

```text
Distance = √((x₂-x₁)² + (y₂-y₁)²)
```

Smaller distance:

```text
More similar
```

Larger distance:

```text
Less similar
```

---

# 7. Linkage 🔗

This is an important technical word.

**Linkage tells the algorithm how to calculate the distance between two clusters.**

For example:

```text
Cluster A       Cluster B

● ●             ● ●
```

How do we calculate the distance between these two groups?

That's where **linkage** comes in.

---

## Types of Linkage

### 1. Single Linkage

Uses the distance between the **closest points** of two clusters.

```text
Cluster A       Cluster B

● ● ----------- ● ●
    closest
```

### 2. Complete Linkage

Uses the distance between the **farthest points**.

### 3. Average Linkage

Uses the **average distance** between points of two clusters.

### 4. Ward Linkage ⭐

Tries to create compact clusters by minimizing the increase in variance.

Common Python example:

```python
linkage="ward"
```

### Easy memory:

```text
Single   → Closest
Complete → Farthest
Average  → Average
Ward     → Compact clusters
```

---

# 8. Dendrogram 🌳

The **Dendrogram** is one of the most important concepts in Hierarchical Clustering.

It is a **tree-like diagram that shows how clusters were merged.**

Example:

```text
             ┌───────────────┐
             │               │
        ┌────┴────┐      ┌───┴───┐
        │         │      │       │
       ┌┴┐       C      ┌┴┐      E
       A B              D F
```

It shows:

> **Which points were merged and at what distance.**

---

# 9. How to choose the number of clusters?

This is where the **Dendrogram** becomes useful.

We draw a horizontal line across the dendrogram.

```text
             ┌───────────────┐
             │               │
        ┌────┴────┐      ┌───┴───┐
        │         │      │       │
       ┌┴┐       C      ┌┴┐      E
       A B              D F

──────────────── ✂ ───────────────
```

The number of branches crossed by the line gives the number of clusters.

For example:

```text
AB
C
DF
E
```

means:

```text
4 clusters
```

### Important word:

**Cutting the dendrogram** = choosing where to separate the hierarchy into final clusters.

# 10. K-Means vs Hierarchical Clustering

| K-Means | Hierarchical |
|---|---|
| Clustering algorithm | Clustering algorithm |
| Unsupervised | Unsupervised |
| Uses centroids | Uses hierarchy |
| Need to choose K | Can use dendrogram |
| Recalculates centroids | Merges/splits clusters |
| No dendrogram | Uses dendrogram |
| Usually faster | Usually slower |
| Good for larger datasets | Good for small/medium datasets |

### Easy difference:

**K-Means:**

> "Tell me how many groups you want."

```text
K = 3
```

Then it creates 3 clusters.

**Hierarchical:**

> "I'll create the complete relationship tree, then you can decide where to cut it."

# 15. When Should We Use Hierarchical Clustering?

Use it when:

✅ You don't have labeled data.

✅ You want to discover natural groups.

✅ You want to understand relationships between groups.

✅ You want a visual hierarchy using a dendrogram.

✅ Dataset is relatively small or medium-sized.

# 16. Limitations

Hierarchical Clustering also has some disadvantages.

### 1. Can be slow

For very large datasets, it can become computationally expensive.

### 2. Sensitive to distance

If your distance measurement is not suitable, clusters may not make sense.

### 3. Sensitive to scaling

If one feature has a much larger range than another, it can dominate the distance.

So scaling is often important.

### 4. Once merged, clusters generally cannot be undone

In agglomerative clustering, once two clusters are merged, that merge is part of the hierarchy.

# 17. Complete Flow

Remember this:

```text
             DATA
               ↓
          Features (X)
               ↓
       Calculate Distance
               ↓
       Find Closest Clusters
               ↓
             Merge
               ↓
       Calculate Distance
               ↓
             Merge
               ↓
          Keep Repeating
               ↓
          DENDROGRAM 🌳
               ↓
          Cut the Tree
               ↓
       Final Clusters
```
