# 📘 Gaussian Mixture Model (GMM) — Easy Notes

## 1. What is GMM?

**GMM = Gaussian Mixture Model**

GMM is an **Unsupervised Machine Learning** algorithm.

### Easy Definition

> GMM groups data points using **probability** and multiple **Gaussian (Normal) distributions**.

---

# 2. Where does GMM come under?

```text
Machine Learning
│
└── Unsupervised Learning
    │
    ├── Clustering
    │   ├── K-Means
    │   ├── Hierarchical Clustering
    │   ├── DBSCAN
    │   └── GMM
    │
    └── Dimensionality Reduction
        └── PCA
```

**Important:** GMM is mainly used for **clustering**.

---

# 3. Main Idea of GMM

Suppose we have a data point:

```text
Customer A
```

GMM can calculate:

```text
Cluster 1 → 80%
Cluster 2 → 20%
```

This means the customer has a higher probability of belonging to Cluster 1.

So GMM is called:

> **Soft Clustering**

---

# 4. K-Means vs GMM

### K-Means

K-Means asks:

> "Which cluster is closest?"

Example:

```text
Customer A
     ↓
Cluster 1
```

The point is assigned to one cluster.

### GMM

GMM asks:

> "What is the probability of this point belonging to each cluster?"

Example:

```text
Customer A

Cluster 1 → 80%
Cluster 2 → 20%
```

---

# 5. Hard Clustering vs Soft Clustering

## K-Means → Hard Clustering

```text
Point
 ↓
Cluster 1
```

One point → one cluster.

## GMM → Soft Clustering

```text
Point
 ↓
Cluster 1 → 70%
Cluster 2 → 30%
```

A point can have a probability for multiple clusters.

---

# 6. Why is it called Gaussian?

**Gaussian** means **Normal Distribution**.

A normal distribution looks like a bell:

```text
              ●
            ● ● ●
          ● ● ● ● ●
        ● ● ● ● ● ● ●
      ● ● ● ● ● ● ● ● ●
```

GMM assumes that the data can be represented using multiple Gaussian distributions.

---

# 7. Why is it called Mixture?

Because GMM combines multiple Gaussian distributions.

For example:

```text
Gaussian 1
     +
Gaussian 2
     +
Gaussian 3
     ↓
   Mixture
```

Therefore:

```text
Gaussian + Mixture + Model
          ↓
          GMM
```

---

# 8. Simple Example

Suppose we have customer data:

```text
Age
Income
Spending Score
```

We want to find customer groups.

GMM might create:

```text
Component 1 → Low Spending
Component 2 → Medium Spending
Component 3 → High Spending
```

# 9. Important Parameter — `n_components`

In GMM, we use:

```python
n_components
```

It tells the model how many Gaussian components we want.

Example:

```python
from sklearn.mixture import GaussianMixture

model = GaussianMixture(
    n_components=3
)
```

Meaning:

```text
Create 3 Gaussian components.
```

### Compare with K-Means

```text
K-Means
→ n_clusters

GMM
→ n_components
```

---

# 10. How GMM Works

GMM commonly uses the **Expectation-Maximization (EM)** algorithm.

Easy steps:

```text
Step 1
↓
Create initial Gaussian components

Step 2
↓
Calculate probability of each point
belonging to each component

Step 3
↓
Update the Gaussian components

Step 4
↓
Repeat the process

Step 5
↓
Stop when the model becomes stable
```

---

# 11. Expectation-Maximization (EM)

EM has two important steps.

## E-Step — Expectation

Calculate:

> "What is the probability that each point belongs to each component?"

Example:

```text
Point A

Cluster 1 → 70%
Cluster 2 → 30%
```

## M-Step — Maximization

Update the Gaussian components using those probabilities.

Then:

```text
E-Step
  ↓
M-Step
  ↓
E-Step
  ↓
M-Step
  ↓
Repeat
```

until the model converges.

# 14. Why Use GMM?

Use GMM when:

- You want **soft clustering**
- Clusters overlap
- You want probabilities
- Data is approximately Gaussian
- K-Means is too rigid
- You want more flexible cluster shapes

# 15. K-Means vs GMM — Important

| Feature | K-Means | GMM |
|---|---|---|
| Learning type | Unsupervised | Unsupervised |
| Task | Clustering | Clustering |
| Clustering | Hard | Soft |
| Main idea | Distance | Probability |
| Main parameter | `n_clusters` | `n_components` |
| Center | Centroid | Gaussian distribution |
| Probability | No | Yes |
| Algorithm | Iterative centroid update | EM |

# 16. GMM vs DBSCAN

### GMM

```text
Probability-based
↓
Soft clustering
↓
Usually assumes Gaussian-shaped distributions
```

### DBSCAN

```text
Density-based
↓
Clusters + Noise
↓
Can find irregular-shaped clusters
```

# 18. Advantages of GMM

✅ Gives probabilities

✅ Supports soft clustering

✅ Can handle overlapping clusters

✅ More flexible than K-Means for some data distributions

# 19. Limitations of GMM

❌ You usually need to choose `n_components`

❌ Results can depend on initialization

❌ Can be sensitive to feature scaling

❌ Works best when the Gaussian assumption is reasonable

❌ Can be affected by outliers

# 22. Important Terms

### Gaussian

Normal/Bell-shaped probability distribution.

### Mixture

Combination of multiple Gaussian distributions.

### Component

One Gaussian distribution inside the mixture.

### `n_components`

Number of Gaussian components.

### Soft Clustering

A point can have probabilities for multiple clusters.

### EM

Expectation-Maximization algorithm used to estimate the Gaussian components.

### `predict_proba()`

Returns the probability of each sample belonging to each component.
