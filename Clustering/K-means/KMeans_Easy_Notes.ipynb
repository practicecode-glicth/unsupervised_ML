# K-Means Clustering — Easy Notes

## 1. What is K-Means?

**K-Means is an Unsupervised Machine Learning algorithm that groups similar data points into clusters.**

```text
Customer Data
     ↓
  K-Means
     ↓
Cluster 1
Cluster 2
Cluster 3
```

K-Means does **not** need a target/answer column.

---

## 2. Why is it called K-Means?

### K
**K = Number of clusters/groups.**

```text
K = 3
↓
3 clusters
```

### Means
The algorithm calculates the **mean/average position** of points in a cluster.

This center is called the **centroid**.

```text
K → Number of clusters
Means → Mean/average → Centroid
```

---

## 3. What is a Centroid?

> **Centroid = the center/average position of the data points in a cluster.**

```text
      ●
   ●  ✖  ●
      ●

✖ = Centroid
```

---

## 4. Do We Decide the Centroid?

We do **not manually choose the final centroids**.

K-Means starts with initial centroids and repeatedly updates them.

```text
Initial centroids
       ↓
Calculate distance
       ↓
Assign points to nearest centroid
       ↓
Calculate new centroids
       ↓
Assign points again
       ↓
Repeat
       ↓
Clusters become stable
```

The algorithm finds a good clustering solution through repeated updates.

---

## 5. How Does K-Means Assign a Point?

K-Means commonly uses **Euclidean Distance**.

Example:

```text
Point A

Distance to Centroid 1 = 2
Distance to Centroid 2 = 8
Distance to Centroid 3 = 12
```

Point A goes to **Cluster 1** because Centroid 1 is closest.

```text
Smaller distance
      ↓
Closer centroid
      ↓
Assign to that cluster
```

---

## 6. Complete K-Means Process

Suppose:

```text
K = 3
```

Then:

```text
1. Choose K = 3
        ↓
2. Start with initial centroids
        ↓
3. Calculate distances
        ↓
4. Assign points to nearest centroid
        ↓
5. Calculate new centroids
        ↓
6. Repeat
        ↓
7. Final/stable clusters
```

## 8. What is the Elbow Method?

The main question is:

> **How do we choose K?**

For example:

```text
K = 2?
K = 3?
K = 4?
K = 5?
```

A common method is the **Elbow Method**.

---

## 9. What is Inertia?

> **Inertia measures how far data points are from their assigned cluster centroids.**

Easy meaning:

> **Inertia tells us how spread out the points are around their centroids.**

```text
Points close to centroid
        ↓
Small distance
        ↓
Lower inertia
```

```text
Points far from centroid
        ↓
Large distance
        ↓
Higher inertia
```

So:

> **Lower inertia = tighter clusters.**

Technically, inertia is the **sum of squared distances** of samples to their closest cluster center.

---

## 10. Why Not Just Choose the Lowest Inertia?

Because increasing K usually decreases inertia.

Example:

```text
K = 1 → 500
K = 2 → 300
K = 3 → 180
K = 4 → 150
K = 5 → 135
K = 6 → 125
```

The improvement becomes smaller after a certain point.

We look for where the curve starts bending:

> **This is the elbow.**

---

## 11. Elbow Method

```text
Inertia
  |
500|●
   | 400|     |   300|    ●
   |     200|         |       ●
150|         ●
   |           ●
100|             ●
   |
   +----------------
      1  2  3  4  5
             K
```

If the curve bends around K = 3, we may choose:

```text
K = 3
```

---

## 12. Elbow Method Steps

```text
Try K = 1
Try K = 2
Try K = 3
Try K = 4
Try K = 5
       ↓
Calculate inertia
       ↓
Plot K vs Inertia
       ↓
Find the elbow
       ↓
Choose a reasonable K
```

---

## 13. K-Means vs Elbow Method

This is very important.

### K-Means

> **Finds the clusters and centroids.**

### Elbow Method

> **Helps us decide a reasonable number of clusters (K).**

Think:

```text
Elbow Method
      ↓
Choose K
      ↓
K-Means
      ↓
Find clusters
```

---

## 14. Does K-Means Have a Target Column?

**No.**

K-Means is **Unsupervised Learning**.

### Supervised:

```text
X + y
 ↓
Model
 ↓
Prediction
```

### K-Means:

```text
X only
 ↓
K-Means
 ↓
Find groups
```

---

## 15. Important Parameter

The main parameter is:

```python
n_clusters
```

Example:

```python
model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)
```

Here:

```text
n_clusters = 3
```

means:

> Create 3 clusters.

Remember:

```text
K = n_clusters
```

---

## 16. Simple Python Example

```python
from sklearn.cluster import KMeans

# Customer data
X = [
    [20, 80],
    [22, 85],
    [25, 82],
    [40, 30],
    [42, 35],
    [45, 32]
]

# Create K-Means model
model = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

# Train the model
model.fit(X)

# Get cluster labels
labels = model.labels_

print(labels)

# Get final centroids
print(model.cluster_centers_)
```

---

## 17. Feature Scaling

K-Means is a **distance-based algorithm**.

Suppose:

```text
Age     → 20–60
Salary  → 20,000–2,00,000
```

Salary has much larger numbers and can strongly affect distance.

So we often use feature scaling.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

model.fit(X_scaled)
```

Remember:

```text
K-Means
   ↓
Uses distance
   ↓
Feature scaling is often important
```

# 🧠 One-Line Definitions

### K-Means
**Groups similar data points into K clusters.**

### Centroid
**The center/mean position of a cluster.**

### Euclidean Distance
**Measures distance between a data point and a centroid.**

### Inertia
**Measures the total squared distance of points from their assigned centroids.**

### Elbow Method
**Helps choose a reasonable K by checking how inertia changes for different K values.**
