# 🟢 DBSCAN — Easy Notes

## 1. What is DBSCAN?

**DBSCAN** stands for:

> **Density-Based Spatial Clustering of Applications with Noise**

DBSCAN is an **Unsupervised Machine Learning** algorithm used for **clustering**.

### Simple meaning

> DBSCAN finds areas where many data points are close together and creates clusters.

It can also identify points that do not belong to any cluster.

Those points are called **Noise / Outliers**.

---

# 2. Simple Example

Imagine customers represented as points:

```text
       ● ● ●
      ● ● ●

                   ●

                        ● ●
                       ● ● ●
```

DBSCAN can find:

```text
Cluster 0       Noise       Cluster 1

 ● ● ●             ●          ● ●
● ● ●                        ● ● ●
```

So DBSCAN can create:

```text
Cluster 0
Cluster 1
Noise
```

---

# 3. Why is it called Density-Based?

**Density** means:

> How many points are present in a particular area.

### High Density

```text
● ● ●
● ● ●
 ● ●
```

Many points are close together.

Therefore:

```text
→ Cluster
```

### Low Density

```text
●


        ●


                 ●
```

Points are far apart.

Therefore:

```text
→ Possible Noise
```

# 5. What is `eps`?

`eps` means:

> **The maximum distance we look around a data point to find its neighbors.**

Think of it as a **radius**.

```text
          ●
       ●  ●  ●
         ●
          ↑
       Data Point

       ← eps →
```

Points inside this distance are considered neighbors.

---

# 6. What is `min_samples`?

`min_samples` means:

> **Minimum number of nearby points required to consider an area dense.**

Example:

```python
min_samples=5
```

means DBSCAN looks for enough nearby points within the `eps` radius.

---

# 7. Core Point ⭐

A **Core Point** is a point that has enough neighboring points within its `eps` distance.

Example:

```text
       ●
    ●  ●  ●
       ●
       ●
```

If the required number of neighbors is reached:

```text
→ Core Point
```

### Easy memory:

> **Core = Lots of nearby points**

---

# 8. Border Point

A **Border Point** does not have enough neighbors to become a Core Point itself.

But it is close to a Core Point.

Example:

```text
       ● ● ●
      ● ● ●
           ●
           ↑
       Border Point
```

The border point can become part of the cluster.

### Easy memory:

> **Border = Near a Core**

---

# 9. Noise Point 🚨

A point becomes **Noise** when it:

- Is not a Core Point
- Is not connected to a Core Point

Example:

```text
● ● ●
● ● ●


                    ●
                    ↑
                  Noise
```

In Python, DBSCAN normally represents noise as:

```text
-1
```

Example:

```python
labels = [0, 0, 0, 1, 1, 1, -1]
```

Here:

```text
0  → Cluster 0
1  → Cluster 1
-1 → Noise
```

---

# 10. Three Important Terms

Remember these three:

```text
Core
Border
Noise
```

### Easy formula:

```text
Core   → Enough neighbors
Border → Near a Core
Noise  → Not connected
```

---

# 11. How DBSCAN Works

The basic process:

```text
                DATA
                  ↓
            Select a Point
                  ↓
       Check eps neighborhood
                  ↓
       Enough nearby points?
             ↙          ↘
           YES           NO
            ↓             ↓
       Core Point     Border/Noise
            ↓
       Create Cluster
            ↓
     Expand the Cluster
            ↓
       Find connected
        dense points
            ↓
          Repeat
```

---

# 12. Does DBSCAN Need K?

### No ❌

This is an important difference from K-Means.

### K-Means:

```python
K = 3
```

You tell the algorithm:

> I want 3 clusters.

### DBSCAN:

You don't directly tell it the number of clusters.

Instead, you give:

```python
eps
min_samples
```

DBSCAN discovers the clusters based on density.

---

# 13. K-Means vs DBSCAN

| K-Means | DBSCAN |
|---|---|
| Unsupervised | Unsupervised |
| Centroid-based | Density-based |
| Need K | Don't directly need K |
| Uses centroids | Uses neighborhoods |
| Sensitive to outliers | Can identify noise |
| Usually every point gets a cluster | Some points can become noise |
| Works well with compact clusters | Can find irregular-shaped clusters |

### Easy memory:

```text
K-Means
→ Centroid

DBSCAN
→ Density
```

---

# 14. DBSCAN vs Hierarchical Clustering

| Hierarchical | DBSCAN |
|---|---|
| Unsupervised | Unsupervised |
| Creates hierarchy | Finds dense regions |
| Uses dendrogram | Uses eps/min_samples |
| Agglomerative commonly used | Density-based |
| Can be computationally expensive | Often efficient |
| Doesn't naturally focus on noise | Can identify noise |

### Easy memory:

```text
Hierarchical → Tree 🌳
DBSCAN       → Density 🟢
```

# 20. Choosing `eps` ⚠️

Choosing a good `eps` value is very important.

### If `eps` is too small:

```text
Few points become neighbors
        ↓
Many points may become Noise
```

### If `eps` is too large:

```text
Many points become neighbors
        ↓
Different clusters may merge
```

So we need a suitable `eps`.

---

# 21. Choosing `eps` Using K-Distance

A common technique is the **K-Distance Graph**.

We calculate the distance to the `k`th nearest neighbor.

Then look for a **knee/elbow point**.

Conceptually:

```text
Distance
   │
   │                 /
   │              /
   │           /
   │        /
   │______/
   │
   └──────────────────
             ↑
           Knee
```

The knee can be used as a starting point for choosing `eps`.

---

# 22. Choosing `min_samples`

`min_samples` controls how many nearby points are needed to create a dense region.

Example:

```python
min_samples=5
```

means the algorithm requires enough nearby observations for density.

### If `min_samples` is increased:

```text
More strict
    ↓
Fewer areas considered dense
```

### If `min_samples` is decreased:

```text
Less strict
    ↓
More areas may become clusters
```

# 24. Advantages of DBSCAN

### ✅ 1. No need to specify K

You don't directly tell it the number of clusters.

### ✅ 2. Finds irregular shapes

It is not limited to round/compact clusters.

### ✅ 3. Detects noise

Noise points can be identified as:

```text
-1
```

### ✅ 4. Useful for outlier detection

It can naturally identify isolated observations.

---

# 25. Disadvantages of DBSCAN

### ❌ 1. Choosing `eps` can be difficult

A poor value can produce bad clusters.

### ❌ 2. Different densities can be difficult

If one cluster is very dense and another is very spread out, one `eps` value may not work well for both.

### ❌ 3. Sensitive to feature scaling

Because it uses distance.

### ❌ 4. High-dimensional data can be difficult

Distance becomes less meaningful as the number of dimensions becomes very large.
