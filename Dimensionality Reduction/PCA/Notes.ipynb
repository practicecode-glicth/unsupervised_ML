**PCA - Principal Component Analysis**

_Simple Notes for Beginners_

# 1\. What is PCA?

PCA stands for Principal Component Analysis. It is an unsupervised learning technique used for dimensionality reduction.

Dimensionality reduction means reducing the number of features in a dataset while trying to keep as much important information as possible.

# 2\. Simple Example

Suppose a dataset has 10 features such as Age, Salary, Experience, Projects, Skills, Attendance, Performance, and Training Hours.

We can use PCA to transform those 10 features into fewer new features:

10 original features → PCA → 2 Principal Components

The two new features are usually called PC1 and PC2.

# 3\. What does PCA actually do?

PCA does not simply divide similar features into groups.

PCA creates new features called Principal Components by combining information from the original features.

For example, Math, Physics, and Chemistry may contain related information. PCA can combine information from these original features into a new component such as PC1.

# 4\. Do we give the number of components?

Yes. We normally tell PCA how many principal components we want using n_components.

# 5\. What does PCA decide automatically?

We choose how many components we want, but PCA automatically calculates how the original features should be combined to create those components.

We do not normally tell PCA which exact columns should become PC1 or PC2.

PCA finds directions that capture the greatest amount of variation in the data.

# 6\. Easy way to remember

We decide HOW MANY components we want; PCA decides HOW to create them.

# 7\. PC1 and PC2

PC1 (Principal Component 1) captures the greatest amount of variation in the data.

PC2 captures the next greatest amount of variation while being orthogonal to PC1.

PC3 captures the next amount, and so on.

# 8\. Why use PCA?

Reduce the number of features.

Make high-dimensional data easier to visualize.

Reduce dataset complexity.

Reduce some redundancy caused by correlated features.

Make some machine learning workflows faster.

# 9\. When should we use PCA?

When a dataset has many numerical features.

When you want to visualize high-dimensional data in 2D or 3D.

When many features contain overlapping or redundant information.

When you want a smaller set of transformed features for another ML model.

# 10\. When should we be careful with PCA?

PCA transforms the original features, so the new components can be harder to interpret.

PCA is sensitive to feature scale, so standardization is commonly performed before PCA.

PCA is mainly designed for numerical data.

Reducing dimensions can discard some information.

# 11\. Basic PCA Workflow

1\. Collect the dataset.

2\. Select numerical features.

3\. Clean the data.

4\. Scale or standardize the features when appropriate.

5\. Choose the number of components.

6\. Fit PCA on the training data.

7\. Transform the data into principal components.

8\. Use the transformed data for visualization or another ML task.

# 12\. Simple Python Example

from sklearn.preprocessing import StandardScaler

from sklearn.decomposition import PCA

\# Example data

X = \[\[90, 92, 88\], \[85, 87, 84\], \[40, 45, 42\], \[50, 52, 48\]\]

\# Standardize the features

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

\# Ask PCA to create 2 components

pca = PCA(n_components=2)

\# Transform the original data into 2 components

X_pca = pca.fit_transform(X_scaled)

print(X_pca)

# 13\. PCA vs Clustering

PCA → Dimensionality reduction → creates PC1, PC2, etc.

K-Means → Clustering → creates groups such as Cluster 1, Cluster 2, etc.

PCA reduces dimensions; clustering groups similar observations.

# 14\. Most Important Point

PCA does not simply select or divide the original columns. It creates new combinations of the original features.

You choose n_components. PCA automatically calculates the principal components.
